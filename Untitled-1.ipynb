git log: ultimas versoes e IDs delas (na visão do commit HEAD (ou seja, o commit que vc está vendo agora) (Leia o de baixo pra entender melhor))

git checkout "ID": dá rollback pra o commit indicado. Além disso o log tbm dá rollback, daí pra mostrar os commits mais recentes eh so dar log --all



pra trocar o HEAD (ou seja, o ponteiro de qual commit tu tá) sem alterar o texto, só trocar o ponteiro mesmo, use git switch
mas se tu quiser puxar só um arquivo de um outro commit, lança:
git restore --source=<ID-do-commit> arquivo.txt

se vc quiser viajar no tempo e ler as coisas antigas....
dá "git stash" com o HEAD na branch q vc ta mexendo
dai c pode sair dando checkout nos commits, que vai voltar todo o texto daquela epoca sombria

pra evitar conflito da stash com os arquivos antigos, antes de dar "stash pop" volta pra branch q tu tava
