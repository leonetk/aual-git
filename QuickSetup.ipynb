Quick Setup - If you've done this kind of thing before\
**Set up in Desktop** or **HTTPS|_SSH_** `https://github.com/leonetk/aual-git.git`

...or create a new repository on the command line
```
echo "# aual-git" >> README.md (1) 
git add README.md
git commit -m "first commit"
git branch -M main (2)
git remote add origin https://github.com/leonetk/aual-git.git
git push -u origin main (3) 
```

*(1)(echo imprime o "# aual-git", e >> redireciona o texto para o final do arquivo a seguir. No caso, README.md.)\
(2)(-M significa renomeação forçada, enquanto -m significa renomeação.)\
(3)(-u significa --set-upstream. Nesse caso, registra que a branch em que estou (main) agora rastreia a branch (origin/main). Isso faz com que os pushes desta branch sejam padronizados para irem pra lá.)*

...or push an existing repository from the command line
```
git remote add origin https://github.com/leonetk/aual-git.git
git branch -M main
git push -u origin main
```
